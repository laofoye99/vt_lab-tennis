```python
import cv2
import threading

def display_video():
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print(f"Error: Could not open video stream 0")
        return
    
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        # Display the resulting frame
        cv2.imshow(f'Video Stream 0', frame)
        
        # Break the loop on 'q' key press
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
        time.sleep(0.04)
    
    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    # Create threads for each video stream
    thread1 = threading.Thread(target=display_video)
    thread2 = threading.Thread(target=display_video)
    
    # Start threads
    thread1.start()
    thread2.start()
    
    # Wait for all threads to complete
    thread1.join()
    thread2.join()



```

```python
import cv2
from multiprocessing import Process, Queue
import time
from queue import Full


def run_person_detector(queue):
    while True:
        frame = queue.get()
        if frame is None:
            break
        cv2.imshow('Person Detector', frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
        
def run_tennis_detector(queue):
    while True:
        frame = queue.get()
        if frame is None:
            break
        cv2.imshow('Tennis Detector', frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
        
if __name__ == "__main__":
    person_queue = Queue(maxsize=25)
    tennis_queue = Queue(maxsize=25)
    
    person_process = Process(target=run_person_detector, args=(person_queue,))
    tennis_process = Process(target=run_tennis_detector, args=(tennis_queue,))
    
    person_process.start()
    tennis_process.start()
    
    cap = cv2.VideoCapture('/home/max/Desktop/tennis/tennis-v5/data/InputVideos/input.mp4')
    try:
        while cap.isOpened():
            ret, frame = cap.read()
            if not ret:
                break
            frame_copy = frame.copy()
            try:
                person_queue.put(frame_copy, block=False)
                tennis_queue.put(frame_copy, block=False)
            except Full:
                print("Queue is full, skipping frame")
                continue
            
            time.sleep(0.04)
    finally:
        cap.release()
        
        person_queue.put(None)
        tennis_queue.put(None)
        
        person_process.join()
        tennis_process.join()
        
        print("All processes joined")
```

```python
import numpy as np
from multiprocessing import Process, Queue
from queue import Full
import time
from multiprocessing.shared_memory import SharedMemory

def producer(q, use_shared_memory):
    frame = np.zeros((1080,1920,3), dtype=np.uint8)
    for _ in range(10000):
        try:
            if use_shared_memory:
                shm = SharedMemory(create=True, size=frame.nbytes)
                shm_frame = np.ndarray(frame.shape, dtype=frame.dtype, buffer=shm.buf)
                shm_frame[:] = frame[:]
                q.put(shm.name)
            else:
                q.put(frame.copy())
        except Full:
            pass
    q.put(None)

def consumer(q, with_print, use_shared_memory):
    count = 0
    start = time.perf_counter()
    while True:
        item = q.get()
        if item is None: break
        
        if use_shared_memory:
            shm = SharedMemory(name=item)
            frame = np.ndarray((1080,1920,3), dtype=np.uint8, buffer=shm.buf)
            shm.close()
            shm.unlink()
        
        if with_print:
            print(f"Processing...")  # 替换为实际输出操作
        
        count += 1
    print(f"Throughput: {count/(time.perf_counter()-start):.2f} fps")

if __name__ == "__main__":
    methods = [
        ("原生+打印", False, True),
        ("共享内存+批处理", True, False)
    ]
    
    for desc, use_shm, with_print in methods:
        q = Queue(maxsize=5)
        p = Process(target=producer, args=(q, use_shm))
        c = Process(target=consumer, args=(q, with_print, use_shm))
        p.start()
        c.start()
        p.join()
        c.join()
```
